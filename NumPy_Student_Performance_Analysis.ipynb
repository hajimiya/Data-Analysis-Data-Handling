## Dataset Creation

Create a NumPy array containing student names, subject names, and marks. This is the only setup section; the remaining sections are question → solution.

```python
import numpy as np

students = np.array([
    "Aarav", "Diya", "Kabir", "Meera", "Ishaan",
    "Anaya", "Rehan", "Sana", "Vihaan", "Zoya"
])

subjects = np.array(["Maths", "Science", "English", "Computer"])

marks = np.array([
    [88, 91, 78, 95],
    [76, 84, 89, 92],
    [59, 63, 71, 68],
    [95, 97, 93, 99],
    [42, 55, 48, 61],
    [67, 72, 81, 79],
    [31, 46, 39, 52],
    [83, 79, 88, 86],
    [54, 49, 58, 63],
    [90, 86, 94, 91]
])
```

## Maximum and Minimum

### 8. Find the maximum marks in Maths.

```python
np.max(marks[:, 0])
```

### 9. Find the minimum marks in Maths.

```python
np.min(marks[:, 0])
```

### 10. Find the maximum marks in each subject.

```python
np.max(marks, axis=0)
```

### 11. Find the minimum marks in each subject.

```python
np.min(marks, axis=0)
```

### 12. Find the highest mark in the complete dataset.

```python
np.max(marks)
```

### 13. Find the lowest mark in the complete dataset.

```python
np.min(marks)
```

### 14. Find the student who obtained the highest individual mark.

```python
students[np.unravel_index(np.argmax(marks), marks.shape)[0]]
```

### 15. Find the subject in which the highest individual mark was obtained.

```python
subjects[np.unravel_index(np.argmax(marks), marks.shape)[1]]
```

## Averages and Statistics

### 16. Find the average marks in Maths.

```python
np.mean(marks[:, 0])
```

### 17. Find the average marks in each subject.

```python
np.mean(marks, axis=0)
```

### 18. Find the average marks of each student.

```python
np.mean(marks, axis=1)
```

### 19. Find the median marks in each subject.

```python
np.median(marks, axis=0)
```

### 20. Find the standard deviation of marks in each subject.

```python
np.std(marks, axis=0)
```

### 21. Find the total marks obtained by each student.

```python
np.sum(marks, axis=1)
```

### 22. Find the overall average of all marks.

```python
np.mean(marks)
```

## Filtering and Boolean Masking

### 23. Find students who scored at least 80 marks in Maths.

```python
students[marks[:, 0] >= 80]
```

### 24. Find students who scored below 40 in at least one subject.

```python
students[np.any(marks < 40, axis=1)]
```

### 25. Find students who scored 40 or more in every subject.

```python
students[np.all(marks >= 40, axis=1)]
```

### 26. Find the number of students who passed every subject, using 40 as the pass mark.

```python
np.sum(np.all(marks >= 40, axis=1))
```

### 27. Find the pass percentage for each subject, using 40 as the pass mark.

```python
np.mean(marks >= 40, axis=0) * 100
```

### 28. Find all marks that are 90 or higher.

```python
marks[marks >= 90]
```

## Sorting and Ranking

### 29. Find the average marks of students, sorted from highest to lowest.

```python
np.sort(np.mean(marks, axis=1))[::-1]
```

### 30. Display student names in descending order of their average marks.

```python
students[np.argsort(np.mean(marks, axis=1))[::-1]]
```

### 31. Find the name of the student with the highest average marks.

```python
students[np.argmax(np.mean(marks, axis=1))]
```

### 32. Find the name of the student with the lowest average marks.

```python
students[np.argmin(np.mean(marks, axis=1))]
```

## Indexing, Slicing, and Broadcasting

### 33. Select the Maths and Science marks of all students.

```python
marks[:, :2]
```

### 34. Select marks for the first five students.

```python
marks[:5]
```

### 35. Calculate how far each mark is above or below its subject average.

```python
marks - np.mean(marks, axis=0)
```

### 36. Count how many students scored above the average in each subject.

```python
np.sum(marks > np.mean(marks, axis=0), axis=0)
```

### 37. Find the average, highest mark, and lowest mark for the complete dataset.

```python
np.array([np.mean(marks), np.max(marks), np.min(marks)])
```
