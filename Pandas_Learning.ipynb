## 8. GroupBy and Aggregation

### Calculate total quantity and average price by category.

```python
df.groupby("category").agg(total_quantity=("quantity", "sum"), average_price=("unit_price", "mean"))
```

### Calculate order count and total revenue by city.

```python
df.assign(revenue=df["quantity"] * df["unit_price"]).groupby("city").agg(order_count=("order_id", "count"), total_revenue=("revenue", "sum"))
```

### Calculate total quantity by city and category.

```python
df.groupby(["city", "category"])["quantity"].sum()
```

### Find the category with the highest total quantity.

```python
df.groupby("category")["quantity"].sum().idxmax()
```

### Calculate min, max, mean, and median unit price by category.

```python
df.groupby("category")["unit_price"].agg(["min", "max", "mean", "median"])
```

## 9. Merge, Join, and Concatenation

### Merge orders with a customer membership table.

```python
customer_info = pd.DataFrame({
    "customer": ["Asha", "Ravi", "Meera", "Kabir", "Nina", "Ira"],
    "membership": ["Gold", "Silver", "Gold", "Bronze", "Silver", "Gold"]
})
df.merge(customer_info, on="customer", how="left")
```

### Join two DataFrames using their index.

```python
left = pd.DataFrame({"customer": ["Asha", "Ravi"]}, index=[1, 2])
right = pd.DataFrame({"membership": ["Gold", "Silver"]}, index=[1, 2])
left.join(right)
```

### Concatenate DataFrames vertically.

```python
jan = pd.DataFrame({"month": ["Jan"], "sales": [1200]})
feb = pd.DataFrame({"month": ["Feb"], "sales": [1500]})
pd.concat([jan, feb], ignore_index=True)
```

### Concatenate DataFrames horizontally.

```python
pd.concat([left.reset_index(drop=True), right.reset_index(drop=True)], axis=1)
```

## 10. Date/Time Data

### Convert order_date to Pandas datetime.

```python
df.assign(order_date=pd.to_datetime(df["order_date"]))
```

### Extract year, month, and day.

```python
dates = pd.to_datetime(df["order_date"])
pd.DataFrame({"year": dates.dt.year, "month": dates.dt.month, "day": dates.dt.day})
```

### Filter orders placed during February 2025.

```python
df[pd.to_datetime(df["order_date"]).dt.month == 2]
```

### Find days since the earliest order.

```python
(pd.to_datetime(df["order_date"]) - pd.to_datetime(df["order_date"]).min()).dt.days
```

### Resample revenue into monthly totals.

```python
daily_revenue = df.assign(
    order_date=pd.to_datetime(df["order_date"]),
    revenue=df["quantity"] * df["unit_price"]
).set_index("order_date")["revenue"]
daily_revenue.resample("MS").sum()
```

## 12. Apply, Map, and Vectorized Operations

### Calculate revenue using vectorized multiplication.

```python
df.assign(revenue=df["quantity"] * df["unit_price"])
```

### Map categories to department codes.

```python
df["category"].map({"Electronics": "ELEC", "Home": "HOME", "Books": "BOOKS"})
```

### Apply a 10% price increase to each unit price.

```python
df["unit_price"].apply(lambda price: price * 1.10)
```

### Label orders as Large or Small using vectorized conditional logic.

```python
df.assign(order_size=np.where(df["quantity"] >= 3, "Large", "Small"))
```

### Calculate revenue and flag orders above 1000.

```python
df.assign(revenue=df["quantity"] * df["unit_price"], high_value=(df["quantity"] * df["unit_price"]) > 1000)
```

## 13. Export Cleaned Data

### Create cleaned data with unique order IDs, filled missing values, revenue, and datetime.

```python
cleaned_df = (
    df.drop_duplicates(subset=["order_id"], keep="first")
      .assign(
          rating=lambda data: data["rating"].fillna(data["rating"].median()),
          discount_code=lambda data: data["discount_code"].fillna("NO_DISCOUNT"),
          revenue=lambda data: data["quantity"] * data["unit_price"],
          order_date=lambda data: pd.to_datetime(data["order_date"])
      )
)
cleaned_df
```

### Export cleaned data to CSV.

```python
cleaned_df.to_csv("cleaned_sales_data.csv", index=False)
```

### Export cleaned data to Excel.

```python
cleaned_df.to_excel("cleaned_sales_data.xlsx", index=False)
```

### Export cleaned data to JSON records.

```python
cleaned_df.to_json("cleaned_sales_data.json", orient="records", date_format="iso", indent=2)
```
