## 1. Arrays and Array Creation

### Create a one-dimensional NumPy array.

```python
import numpy as np

scores = np.array([72, 85, 91, 64, 78])
scores
```

### Create a two-dimensional array of student marks.

```python
marks = np.array([[78, 85, 91], [66, 74, 80], [92, 89, 95], [55, 68, 72]])
marks
```

### Create arrays of zeros, ones, and a range of values.

```python
np.zeros((2, 3)), np.ones((2, 3)), np.arange(0, 13, 2)
```

### Create evenly spaced values between 0 and 1.

```python
np.linspace(0, 1, 6)
```

## 7. Aggregations and Axis Operations

### Find the total, mean, median, minimum, and maximum of all marks.

```python
np.sum(marks), np.mean(marks), np.median(marks), np.min(marks), np.max(marks)
```

### Find the average mark for each student.

```python
np.mean(marks, axis=1)
```

### Find the average mark for each subject.

```python
np.mean(marks, axis=0)
```

### Find the standard deviation for each subject.

```python
np.std(marks, axis=0)
```

### Find the index of the highest mark in the complete array.

```python
np.unravel_index(np.argmax(marks), marks.shape)
```

### Count how many marks are 80 or higher in each subject.

```python
np.sum(marks >= 80, axis=0)
```

## 8. Boolean Masks

### Find all marks between 70 and 89 inclusive.

```python
marks[(marks >= 70) & (marks <= 89)]
```

### Find students who scored at least 80 in every subject.

```python
np.all(marks >= 80, axis=1)
```

### Find students who scored below 40 in at least one subject.

```python
np.any(marks < 40, axis=1)
```

### Replace marks below 40 with 40 in a new array.

```python
np.where(marks < 40, 40, marks)
```

## 9. Random Number Generation

### Generate reproducible random integers between 1 and 100.

```python
rng = np.random.default_rng(42)
rng.integers(1, 101, size=10)
```

### Generate a reproducible 3 × 4 matrix of normally distributed values.

```python
rng = np.random.default_rng(42)
rng.normal(loc=50, scale=10, size=(3, 4)).round(2)
```

### Randomly select three values from an array without replacement.

```python
rng = np.random.default_rng(42)
rng.choice(np.arange(1, 11), size=3, replace=False)
```

### Shuffle a copy of the scores reproducibly.

```python
rng = np.random.default_rng(42)
shuffled_scores = scores.copy()
rng.shuffle(shuffled_scores)
shuffled_scores
```

## 10. Linear Algebra Functions

### Calculate the dot product of two vectors.

```python
np.dot(np.array([1, 2, 3]), np.array([4, 5, 6]))
```

### Multiply two matrices.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
A @ B
```

### Find the determinant of a square matrix.

```python
np.linalg.det(A)
```

### Find the inverse of an invertible square matrix.

```python
np.linalg.inv(A)
```

### Find the eigenvalues of a square matrix.

```python
np.linalg.eigvals(A)
```

### Solve a system of linear equations: 2x + y = 5 and x + 3y = 6.

```python
np.linalg.solve(np.array([[2, 1], [1, 3]]), np.array([5, 6]))
```

## 11. Array Copying vs Views

### Create a view and a copy of an array.

```python
original = np.array([10, 20, 30, 40])
view = original.view()
copy = original.copy()
view, copy
```

### Change the view and inspect the original array.

```python
view[0] = 999
original
```

### Change the copy and inspect the original array.

```python
copy[1] = 888
original
```

### Check whether two arrays share memory.

```python
np.shares_memory(original, view), np.shares_memory(original, copy)
```

### Reshape an array as a view where possible and check memory sharing.

```python
base = np.arange(12)
reshaped = base.reshape(3, 4)
np.shares_memory(base, reshaped)
```
